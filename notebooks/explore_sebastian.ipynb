```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy
import collections
import pathlib as Path
from tqdm import tqdm
from torch.utils.data import Dataset
from torch.utils.data import DataLoader
from torch import nn
#Given kaggle dependecies are installed we can install the dataset
if not os.listdir('../data/'):
    dataset = "aanhari/alexa-dataset"
    !kaggle datasets download -d {dataset} -p ../data/
    !unzip -q ../data/alexa-dataset.zip -d ../data/
    !rm ../data/alexa-dataset.zip
    print("The dataset has been downloaded and unzipped.")
    
    base_dir = "data/alexa"

    global_idx = 1

    for subfolder in os.listdir(base_dir):
        subfolder_path = os.path.join(base_dir, subfolder)

        if os.path.isdir(subfolder_path):
            files = os.listdir(subfolder_path)

            files.sort()

            for file in files:
                old_file = os.path.join(subfolder_path, file)
                new_file = os.path.join(subfolder_path, f"{global_idx}{os.path.splitext(file)[1]}")  # Keep original file extension
                
                os.rename(old_file, new_file)
                print(f"Renamed {old_file} to {new_file}")
                
                global_idx += 1
else:
    print("The data/ folder is not empty.")
```
